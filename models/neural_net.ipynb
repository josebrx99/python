```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Lineal model: One neurone

```python
"""
1. Inicialización
Se crean los parámetros iniciales \(w\) y \(b\).


2. Forward Propagation
La neurona recibe la estatura y calcula el peso estimado.
\(\hat{y}=Xw+b\)


3. Loss Function
Comparamos el peso estimado con el peso real.
\(L=(y-\hat{y})^2\)


4. Backpropagation
Calculamos las derivadas para determinar cómo modificar los parámetros.


5. Gradient Descent
Actualizamos los parámetros para reducir el error.
\(w=w-\alpha dw\)


6. Repetición
Repetimos el proceso durante las épocas de entrenamiento.


7. Predicción
Utilizamos los parámetros aprendidos para estimar el peso de nuevas personas.
"""
```

```python
# Train
X = np.array([150, 155, 160, 165, 170, 175, 180, 185], dtype=float)
y = np.array([45, 53, 50, 70, 50, 70, 80, 70], dtype=float)

X_mean = X.mean()
X_std = X.std()
X = (X - X_mean) / X_std

y_mean = y.mean()
y_std = y.std()
y = (y - y_mean) / y_std

df = pd.DataFrame({"estatura":X, "peso":y})

# Test
X = np.array([], dtype=float)
y = np.array([], dtype=float)

X_mean = X.mean()
X_std = X.std()
X = (X - X_mean) / X_std

y_mean = y.mean()
y_std = y.std()
y = (y - y_mean) / y_std

df = pd.DataFrame({"estatura":X, "peso":y})

df
```

```python
def forward_propagation(X, w, b):
    z = X * w + b
    y_pred = z
    return y_pred

def loss_function(y_real, y_pred):
    loss = np.mean((y_real - y_pred) ** 2)
    return loss

def gradients(X, y, y_pred):
    # Paso 6. Implementar Backpropagation
    # Aquí calcularemos los gradientes de los parámetros. Estamos calculando el gradiente promedio de todo nuestro conjunto de entrenamiento.
    dw = np.mean(2 * (y_pred - y) * X)
    db = np.mean(2 * (y_pred - y))
    return dw, db

def gradient_descent(w, b, dw, db, learning_rate):
    # Paso 7. Implementar Gradient Descent
    w = w - learning_rate * dw
    b = b - learning_rate * db
    return w, b
```

```python
# Paso 8. Entrenar la red neuronal
# Inicialización de parámetros
np.random.seed(42)

w = np.random.randn() # Peso
b = 0.0               # Sesgo

# Hiperparámetros
learning_rate = 0.01
epochs = 200


loss_history = []
l_epochs = range(epochs)
for epoch in l_epochs:

    # Forward propagation
    y_pred = forward_propagation(X, w, b)

    # Calcular error
    loss = loss_function(y, y_pred)
    from sklearn.metrics import r2_score
    r2 = r2_score(y, y_pred)

    # Backpropagation
    dw, db = gradients(X, y, y_pred)

    # Actualizar parámetros
    w, b = gradient_descent(w, b, dw, db, learning_rate)

    loss_history.append(loss)

    if epoch % 25 == 0 or epoch == len(l_epochs)-1 or epoch in [2,4,6,8,10]:
        print(f"Epoch: {epoch:>3} | Loss: {loss:.8f} | R²: {r2*100:.2f}% | w: {w:.4f} | b: {b:.4f}")
print("="*120)

print(f"\nR²: {r2*100:.2f}%")

print("\nParameters")
print(f"w: {w:.4f} | b: {b:.10f}")


plt.figure(figsize=(7.5, 3.4))
plt.plot(loss_history)
plt.xlabel("Épocas")
plt.ylabel("MSE")
plt.title("Evolución del error durante el entrenamiento")
plt.grid()
plt.show()
```

```python
# Predict

df["y_pred"] = df["estatura"]*w + b
df["estatura0"] = df["estatura"] * X_std + X_mean
df["peso0"] = df["peso"] * y_std + y_mean
df["y_pred0"] = df["y_pred"] * y_std + y_mean

r2 = r2_score(df["peso0"], df["y_pred0"])
print(f"R²: {r2*100:.2f}%")

df
```

```python
plt.figure(figsize=(7.5, 3.7))
plt.scatter(df["estatura0"], df["peso0"], label="Datos reales")
plt.plot(df["estatura0"], df["y_pred0"], label="Predicción de la neurona")
plt.ylabel("X")
plt.title("")
# plt.legend()
plt.grid([])
plt.show()
```

## Lineal model: More than one neurone

```python
# Convertir los vectores en matrices columna
X = X.reshape(-1, 1)

y = y.reshape(-1, 1)

print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)
```

```python
# Arquitectura de la red
input_neurons = 1
hidden_neurons = 4
output_neurons = 1
```

```python
def forward(X, W1, b1, W2, b2):
    # Primera capa: entrada hacia capa oculta
    Z1 = X @ W1 + b1
    # Función de activación
    A1 = np.tanh(Z1)
    # Segunda capa: capa oculta hacia salida
    Z2 = A1 @ W2 + b2
    # Activación lineal de salida
    y_pred = Z2
    return Z1, A1, Z2, y_pred

def mse(y_real, y_pred):
    loss = np.mean((y_real - y_pred) ** 2)
    return loss

def backward(X, y, Z1, A1, y_pred, W2):
    n = X.shape[0]
    # Gradiente de la salida
    dZ2 = (2 / n) * (y_pred - y)
    # Gradientes de la segunda capa
    dW2 = A1.T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)
    # Propagar gradientes hacia la capa oculta
    dA1 = dZ2 @ W2.T
    # Derivada de tanh
    dZ1 = dA1 * (1 - A1 ** 2)
    # Gradientes de la primera capa
    dW1 = X.T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)
    return dW1, db1, dW2, db2

def update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, learning_rate):
    W1 = W1 - learning_rate * dW1
    b1 = b1 - learning_rate * db1
    W2 = W2 - learning_rate * dW2
    b2 = b2 - learning_rate * db2
    return W1, b1, W2, b2
```

```python
# Hiperparámetros
learning_rate = 0.03

np.random.seed(42)

# Pesos de entrada a capa oculta
W1 = np.random.randn(input_neurons, hidden_neurons) * 0.5

# Sesgos de capa oculta
b1 = np.zeros((1, hidden_neurons))

# Pesos de capa oculta a salida
W2 = np.random.randn(hidden_neurons, output_neurons) * 0.5

# Sesgo de salida
b2 = np.zeros((1, output_neurons))

epochs = 10000

# Historial del error
loss_history = []

l_epochs = range(epochs)
for epoch in l_epochs:

    # Forward propagation
    Z1, A1, Z2, y_pred = forward(X, W1, b1, W2, b2)

    # Calcular pérdida
    loss = mse(y, y_pred)
    from sklearn.metrics import r2_score
    r2 = r2_score(y, y_pred)

    # Backpropagation
    dW1, db1, dW2, db2 = backward(X, y, Z1, A1, y_pred, W2)

    # Actualizar parámetros
    W1, b1, W2, b2 = update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, learning_rate)

    # Guardar pérdida
    loss_history.append(loss)

    if epoch % 1000 == 0 or epoch == len(l_epochs)-1 or epoch in [2,4,6,8,10]:
        print(f"Epoch: {epoch:>5} | Loss: {loss:.8f} | R²: {r2*100:.2f}% | w: {w:.4f} | b: {b:.4f}")
print("="*120)

print(f"\nR²: {r2*100:.2f}%")


plt.figure(figsize=(7.5, 3.4))
plt.plot(loss_history)
plt.xlabel("Épocas")
plt.ylabel("MSE")
plt.title("Entrenamiento de la red neuronal multicapa")
plt.grid()
plt.show()
```

```python
def predict(height):

    # Convertir entrada en matriz
    height = np.array([[height]], dtype=float)

    # Normalizar estatura
    height_norm = (height - X_mean) / X_std

    # Forward propagation
    Z1, A1, Z2, prediction_norm = forward(height_norm, W1, b1, W2, b2)

    # Desnormalizar predicción
    prediction = prediction_norm * y_std + y_mean

    return prediction.item()
```

```python
# Predict

df["estatura0"] = df["estatura"] * X_std + X_mean
df["y_pred"] = [predict(x) for x in df["estatura0"]]
df["peso0"] = df["peso"] * y_std + y_mean

r2 = r2_score(df["peso0"], df["y_pred"])
print(f"R²: {r2*100:.2f}%")

df
```

```python
plt.figure(figsize=(7.5, 3.7))
plt.scatter(df["estatura0"], df["peso0"], label="Datos reales")
plt.plot(df["estatura0"], df["y_pred0"], label="Predicción de la neurona")
plt.ylabel("X")
plt.title("")
# plt.legend()
plt.grid([])
plt.show()
```
